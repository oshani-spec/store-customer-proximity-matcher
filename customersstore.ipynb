```python
from google.colab import files

print("Upload customers.csv")
customers_file = files.upload()

print("Upload stores.csv")
stores_file = files.upload()
```

```python
import pandas as pd

customers = pd.read_csv(list(customers_file.keys())[0])
stores = pd.read_csv(list(stores_file.keys())[0])

print(customers.head())
print(stores.head())
```

```python
from geopy.geocoders import Nominatim
import time

geolocator = Nominatim(user_agent="geo_project")

def geocode(address):
    try:
        location = geolocator.geocode(address)
        if location:
            return location.latitude, location.longitude
    except:
        return None, None
```

```python
customers['lat'] = None
customers['lon'] = None

for i, row in customers.iterrows():
    lat, lon = geocode(row['Address'])
    customers.at[i, 'lat'] = lat
    customers.at[i, 'lon'] = lon
    print(row['Address'], lat, lon)
    time.sleep(1)
```

```python
stores['lat'] = None
stores['lon'] = None

for i, row in stores.iterrows():
    # Call geocode and store the result
    geo_result = geocode(row['Address'])

    # Check if geo_result is None before attempting to unpack
    if geo_result is not None:
        lat, lon = geo_result
    else:
        # If geocode returned None, assign None to lat and lon explicitly
        lat, lon = None, None

    stores.at[i, 'lat'] = lat
    stores.at[i, 'lon'] = lon
    print(row['Address'], lat, lon)
    time.sleep(1)
```

```python
results_df = customers # Assign customers DataFrame to results_df to enable download
results_df.to_csv("output.csv", index=False)

from google.colab import files
files.download("output.csv")
```

```python
import folium

m = folium.Map(location=[7.8, 80.7], zoom_start=7)

# Customers
for _, row in customers.iterrows():
    if row['lat'] is not None and row['lon'] is not None: # Check for None values
        folium.Marker(
            [row['lat'], row['lon']],
            popup=row['Name'],
            icon=folium.Icon(color='blue')
        ).add_to(m)

# Stores
for _, row in stores.iterrows():
    if row['lat'] is not None and row['lon'] is not None: # Check for None values
        folium.Marker(
            [row['lat'], row['lon']],
            popup=row['StoreName'],
            icon=folium.Icon(color='red')
        ).add_to(m)

m
```

```python
# Save map as HTML file
map_file = "customer_store_map.html"
m.save(map_file)

print("✅ Map saved as:", map_file)

# Download the map (Google Colab)
from google.colab import files
files.download(map_file)

print("📥 Download started!")

```
